```python
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv()) # read local .env file
```

```python
from openai import OpenAI
import json

client = OpenAI()

Model = "gpt-4.1-nano"
```

```python
# Step 1: 첫 번째 질문 처리
initial_response = client.responses.create(
    model=Model,
    input=[{"role": "user", "content": "요즘 소화가 잘 안돼요. 어떻게 해야 할까요?"}],
)

print("\n 첫 번째 응답:")
print(initial_response.output_text)

# Step 2: 이전 대화 상태를 반영한 두 번째 질문
follow_up_response = client.responses.create(
    model=Model,
    previous_response_id=initial_response.id,  # 대화 흐름 유지 핵심
    input=[{"role": "user", "content": "그럼 매운 음식은 피해야 하나요?"}],
    instructions="사용자의 질문 외에 추가된 답변이 없도록 하세요. 답변 마지막에 '다른 증상은 없으신가요?'라는 질문으로 마무리하세요."
)

print("\n 두 번째 응답:")
print(follow_up_response.output_text)
```
