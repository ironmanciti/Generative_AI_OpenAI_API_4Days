```python
# 실습: Agent SDK – 분류, 도구, 핸드오프, 가드레일 통합 예제

from agents import Agent, Runner, function_tool, InputGuardrail, GuardrailFunctionOutput
from dotenv import load_dotenv, find_dotenv
from pydantic import BaseModel
import asyncio
import re

load_dotenv(find_dotenv())
model = "gpt-4o-mini"

### 1. 함수 도구 정의
@function_tool
def calculate_area(length: float, width: float) -> float:
    """
    직사각형의 넓이를 계산합니다. 길이 x 너비
    """
    return length * width

### 2. 가드레일 검사 함수 정의
class SafetyCheck(BaseModel):
    safe: bool
    reason: str

dangerous_keywords = ["해킹", "폭탄", "암살"]

def contains_dangerous_words(text):
    for word in dangerous_keywords:
        if word in text:
            return word
    return None

async def safety_guardrail(ctx, agent, input_data):
    word = contains_dangerous_words(input_data)
    safe = word is None
    return GuardrailFunctionOutput(
        output_info=SafetyCheck(safe=safe, reason=f"금지어 '{word}' 포함" if not safe else "안전"),
        tripwire_triggered=not safe
    )

### 3. 에이전트 정의
math_agent = Agent(
    name="수학 에이전트",
    instructions="수학 문제를 도와주는 튜터입니다. 제공된 도구를 활용하세요.",
    model=model,
    tools=[calculate_area],
)

general_agent = Agent(
    name="일반 에이전트",
    instructions="수학 질문이 아닌 경우, 친절하게 수학만 도와줄 수 있다고 안내하세요.",
    model=model
)

router_agent = Agent(
    name="분류 에이전트",
    instructions="입력이 수학 문제면 수학 에이전트로, 그렇지 않으면 일반 에이전트로 보내세요.",
    model=model,
    handoffs=[math_agent, general_agent],
    input_guardrails=[InputGuardrail(guardrail_function=safety_guardrail)]
)

### 4. 테스트 함수 정의
async def test_agent(input_text):
    try:
        result = await Runner.run(router_agent, input=input_text)
        print("\n입력:", input_text)
        print("응답:", result.final_output)
    except Exception as e:
        print("\n입력:", input_text)
        print("가드레일에 의해 차단됨:", e)

# 테스트 실행
await test_agent("가로 5, 세로 7인 직사각형의 넓이를 구해주세요.")
await test_agent("어제 뉴스에 나온 해킹 사고에 대해 말해줘")
await test_agent("오늘 점심 뭐 먹을까요?")
```
