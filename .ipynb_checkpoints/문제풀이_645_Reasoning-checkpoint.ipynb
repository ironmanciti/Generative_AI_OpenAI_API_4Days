```python
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv()) # read local .env file
```

```python
prompt = """
다음과 같은 기능을 수행하는 파이썬 함수를 작성하세요:

1. 입력은 문자열이며, 쉼표로 구분된 정수로 이루어져 있습니다. 예: "1, 5, 3, 8, 2"
2. 이 문자열을 정수 리스트로 변환합니다.
3. 숫자들의 평균을 계산합니다. 평균은 소수 둘째 자리까지 반올림합니다.
4. 숫자들의 중앙값(중간값)을 계산합니다.
5. 평균과 중앙값을 포함하는 딕셔너리 형식으로 결과를 반환합니다.
6. 적절한 테스트 코드도 포함하고, 단계마다 주석을 달아주세요.
"""
```

```python
from openai import OpenAI

client = OpenAI()

response = client.responses.create(
    model="o4-mini",
    input=[
        {
            "role": "user",
            "content": prompt,
        }
    ]
)

print(response.output_text)
```
